```python
import pandas as pd
```

```python
#불러올 때마다 주소를 바꿔주면 된다.
trans_json = pd.read_json('make_service_for_me/vq6_47FLOeI.json')
trans_json
```

```python
trans_json['transcript']
```

```python
# 자막 스크립트 결합

combined_text = trans_json['transcript'].str.cat(sep=' ')

# 새로운 DataFrame 생성 (기존의 열들과 합쳐서 생성)
new_df = pd.DataFrame([combined_text], columns=['Combined_Transcript'])
new_df = pd.concat([trans_json[['title', 'upload_date']], new_df], axis=1)

# 0번 인덱스만 출력
final_df = new_df.iloc[:1].reset_index(drop=True)
```
